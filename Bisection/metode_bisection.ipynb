# PENYELESAIAN PERSAMAAN NONLINEAR MENGGUNAKAN METODE BISECTION

## 1. Tujuan
Membuat program Python untuk mencari akar persamaan nonlinear menggunakan metode Bisection.

## 2. Persamaan
f(x) = x³ - x - 2

## 3. Interval Awal
a = 1
b = 2

## 4. Toleransi Error
0,0001

## 5. Algoritma
1. Menentukan fungsi persamaan nonlinear.
2. Menentukan batas bawah (a) dan batas atas (b).
3. Menghitung titik tengah interval.
4. Menghitung nilai fungsi pada titik tengah.
5. Memeriksa tanda nilai fungsi.
6. Memperbarui interval berdasarkan hasil pemeriksaan.
7. Mengulangi proses hingga mencapai toleransi error.
8. Menampilkan hasil akar persamaan.

```python
# Fungsi persamaan nonlinear

def f(x):
    return x**3 - x - 2
```

### Penjelasan Kode 1: Fungsi Persamaan Nonlinear

Pada bagian ini dibuat fungsi bernama `f(x)` untuk mendefinisikan persamaan nonlinear yang akan diselesaikan.

Persamaan yang digunakan adalah f(x) = x³ - x - 2.

- `def f(x):` digunakan untuk membuat fungsi dengan nama f dan menerima nilai x sebagai input.
- `return x**3 - x - 2` digunakan untuk menghitung nilai fungsi berdasarkan persamaan yang telah ditentukan.
- Operator `**3` berarti nilai x dipangkatkan tiga.

Fungsi ini digunakan pada proses perhitungan metode Bisection untuk menentukan nilai fungsi pada setiap titik.

```python
# Menentukan interval awal dan toleransi

a = 1
b = 2
toleransi = 0.0001
maks_iterasi = 100

# Memeriksa apakah interval memenuhi syarat
if f(a) * f(b) >= 0:
    raise ValueError("Interval tidak memenuhi syarat Bisection.")

print("Interval awal:", a, "sampai", b)
print("Toleransi error:", toleransi)
```

### Penjelasan Kode 2: Menentukan Interval Awal dan Toleransi

Pada bagian ini dilakukan penentuan nilai awal yang digunakan dalam proses perhitungan metode Bisection.

- `a = 1` digunakan untuk menentukan batas bawah interval.
- `b = 2` digunakan untuk menentukan batas atas interval.
- `toleransi = 0.0001` digunakan untuk menentukan tingkat ketelitian hasil perhitungan.
- `maks_iterasi = 100` digunakan untuk membatasi jumlah pengulangan perhitungan maksimal sebanyak 100 kali.
- `if f(a) * f(b) >= 0` digunakan untuk memeriksa apakah interval awal memenuhi syarat metode Bisection.
- `raise ValueError` digunakan untuk menampilkan pesan kesalahan jika interval tidak memenuhi syarat.
- `print()` digunakan untuk menampilkan interval awal dan nilai toleransi.

Dengan menentukan interval dan toleransi, program dapat melakukan proses pencarian akar persamaan secara terarah.

```python
# Fungsi untuk menyelesaikan persamaan menggunakan Metode Bisection

def metode_bisection(a, b, toleransi, maks_iterasi):

    print("TABEL ITERASI METODE BISECTION")
    print("-" * 65)
    print("Iterasi\t a\t\t b\t\t c\t\t f(c)")
    print("-" * 65)

    for i in range(1, maks_iterasi + 1):

        # Menghitung titik tengah
        c = (a + b) / 2

        # Menghitung nilai fungsi pada titik tengah
        fc = f(c)

        # Menampilkan hasil iterasi
        print(f"{i}\t {a:.6f}\t {b:.6f}\t {c:.6f}\t {fc:.6f}")

        # Memeriksa kondisi berhenti
        if abs(fc) < toleransi:
            break

        # Memperbarui interval
        if f(a) * fc < 0:
            b = c
        else:
            a = c

    return c
```

### Penjelasan Kode 3: Proses Perhitungan Metode Bisection

Pada bagian ini dibuat fungsi utama bernama `metode_bisection()` yang digunakan untuk mencari akar persamaan nonlinear dengan metode Bisection.

- `def metode_bisection(a, b, toleransi, maks_iterasi):` digunakan untuk membuat fungsi dengan parameter batas bawah, batas atas, toleransi, dan jumlah iterasi maksimal.
- `print()` digunakan untuk menampilkan judul dan kolom tabel iterasi.
- `for i in range(1, maks_iterasi + 1):` digunakan untuk mengulang proses perhitungan mulai dari iterasi pertama sampai batas maksimal.
- `c = (a + b) / 2` digunakan untuk menghitung titik tengah dari interval.
- `fc = f(c)` digunakan untuk menghitung nilai fungsi pada titik tengah.
- `print()` digunakan untuk menampilkan nilai iterasi, a, b, c, dan f(c) ke dalam tabel.
- `if abs(fc) < toleransi:` digunakan untuk memeriksa apakah nilai fungsi sudah memenuhi batas toleransi. Jika terpenuhi, proses perhitungan dihentikan.
- `if f(a) * fc < 0:` digunakan untuk menentukan interval baru berdasarkan perubahan tanda fungsi.
- `b = c` digunakan untuk memperbarui batas atas jika akar berada di antara a dan c.
- `a = c` digunakan untuk memperbarui batas bawah jika akar berada di antara c dan b.
- `return c` digunakan untuk mengembalikan nilai titik tengah terakhir sebagai hasil pendekatan akar persamaan.

Dengan proses tersebut, program akan mencari akar persamaan secara berulang sampai memenuhi toleransi atau mencapai batas maksimal iterasi.
