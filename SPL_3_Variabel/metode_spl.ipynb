## Tujuan Program

Program ini dibuat untuk menyelesaikan sistem persamaan linear tiga variabel menggunakan metode Eliminasi Gauss. Proses penyelesaian dilakukan dengan mengubah matriks augmented menjadi matriks segitiga atas, kemudian menentukan nilai setiap variabel melalui substitusi mundur.

## Algoritma Penyelesaian SPL 3 Variabel

1. Mulai program.
2. Masukkan koefisien dari tiga persamaan linear dan konstanta masing-masing persamaan.
3. Bentuk matriks augmented berdasarkan data yang dimasukkan.
4. Tentukan elemen pivot pada setiap kolom.
5. Lakukan pertukaran baris jika elemen pivot bernilai nol atau lebih kecil dari kandidat pivot lainnya.
6. Lakukan eliminasi Gauss untuk mengubah matriks menjadi bentuk segitiga atas.
7. Lakukan substitusi mundur untuk memperoleh nilai variabel.
8. Tampilkan hasil nilai x, y, dan z.
9. Selesai.

```python

# Fungsi untuk menyelesaikan SPL menggunakan Eliminasi Gauss

def eliminasi_gauss(matriks):
    n = len(matriks)

    # Membuat salinan matriks agar data awal tidak berubah
    A = [baris[:] for baris in matriks]

    # Proses eliminasi maju
    for i in range(n):

        # Menentukan baris dengan nilai pivot terbesar
        pivot = max(range(i, n), key=lambda k: abs(A[k][i]))

        # Memeriksa apakah pivot bernilai nol
        if abs(A[pivot][i]) < 1e-12:
            raise ValueError("Sistem tidak memiliki solusi tunggal.")

        # Menukar baris jika diperlukan
        A[i], A[pivot] = A[pivot], A[i]

        # Menghilangkan elemen di bawah pivot
        for j in range(i + 1, n):
            faktor = A[j][i] / A[i][i]

            for k in range(i, n + 1):
                A[j][k] -= faktor * A[i][k]

    # Proses substitusi mundur
    x = [0.0] * n

    for i in range(n - 1, -1, -1):
        jumlah = sum(A[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (A[i][n] - jumlah) / A[i][i]

    return x
```

```python
# Memasukkan koefisien persamaan ke dalam matriks

matriks = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

# Memanggil fungsi Eliminasi Gauss
hasil = eliminasi_gauss(matriks)

# Menampilkan hasil
print("HASIL PENYELESAIAN SPL 3 VARIABEL")
print("---------------------------------")
print(f"Nilai x = {hasil[0]:.6f}")
print(f"Nilai y = {hasil[1]:.6f}")
print(f"Nilai z = {hasil[2]:.6f}")
```

## PENJELASAN KODE PROGRAM ELIMINASI GAUSS

### 1. Membuat Fungsi Eliminasi Gauss
Kode `def eliminasi_gauss(matriks):` digunakan untuk membuat fungsi yang bertugas menyelesaikan sistem persamaan linear menggunakan metode Eliminasi Gauss.

### 2. Menentukan Jumlah Persamaan
Kode `n = len(matriks)` digunakan untuk menghitung jumlah baris atau persamaan yang terdapat di dalam matriks.

### 3. Membuat Salinan Matriks
Kode `A = [baris[:] for baris in matriks]` digunakan untuk menyalin matriks agar data matriks awal tidak berubah selama proses perhitungan.

### 4. Proses Eliminasi Maju
Bagian ini digunakan untuk mengubah matriks menjadi bentuk segitiga atas dengan menghilangkan nilai di bawah elemen pivot.

### 5. Menentukan Pivot
Kode `pivot = max(...)` digunakan untuk mencari nilai pivot terbesar berdasarkan nilai absolut pada kolom yang sedang diproses.

### 6. Menukar Baris
Kode `A[i], A[pivot] = A[pivot], A[i]` digunakan untuk menukar posisi baris apabila diperlukan agar proses perhitungan lebih stabil.

### 7. Menghilangkan Elemen di Bawah Pivot
Proses ini menggunakan faktor eliminasi untuk membuat nilai di bawah pivot menjadi nol.

### 8. Proses Substitusi Mundur
Tahap ini digunakan untuk menghitung nilai setiap variabel mulai dari variabel terakhir hingga variabel pertama.

### 9. Mengembalikan Hasil
Kode `return x` digunakan untuk mengembalikan hasil perhitungan berupa nilai variabel yang telah diperoleh.
